```python
# python 语法

# 数据类型 
# 整形 INTEGER / int
int = -9
print(int)

# 浮点数 FLOAT / float
float1 = 3.14
float2 = .1
float3 = 1.
print(type(float))
print(float1, float2, float3)

# 布尔数据类型 BOOLEAN / bool
bool1 = True # 首字母必须大写
bool2 = False
print(type(bool1))
print(bool1)
print(type(bool2))
print(bool2)

# 字符串类型 STRING / str
str1 = 'hello world'
print(type(str1))
print(str1)
```
