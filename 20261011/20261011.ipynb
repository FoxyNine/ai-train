```python
# python 语法

# 数据类型 
# 整形 INTEGER / int
int = -9
print(int)

# 浮点数 FLOAT / float
float1 = 3.14
float2 = .1
float3 = 1.
print(type(float))
print(float1, float2, float3)

# 布尔数据类型 BOOLEAN / bool
bool1 = True # 首字母必须大写
bool2 = False
print(type(bool1))
print(bool1)
print(type(bool2))
print(bool2)

# 字符串类型 STRING / str
str1 = 'hello world'
print(type(str1))
print(str1)
```

```python
# 输入: input函数, 括号内写输入的提示词, 默认返回的数据类型是str
input1 = input("请输入内容：")
print("你输入的内容是：", input1)

age = int(input("请输入你的年龄：")) # 直接在input函数外面套一个int()函数, 就可以将输入的字符串类型转换为整形
print("数据类型是:", type(age), "你输入的年龄是：", age)
```

```python
# 算数运算符 + - * / // % **
a = 10
b = 3
print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b) # 除法, 返回浮点数
print("a // b =", a // b) # 整除, 返回整形
print("a % b =", a % b) # 取余数
print("a ** b =", a ** b) # 幂运算, 返回整形
print()

# 比较运算操作 == != > < >= <=
print("a == b:", a == b)
print("a != b:", a != b)
print("a > b:", a > b)
print("a < b:", a < b)
print("a >= b:", a >= b)
print("a <= b:", a <= b)
print()

# 逻辑运算符 and or not
x = True
y = False
print("x and y:", x and y)
print("x or y:", x or y)
print("not x:", not x)
```

```python
# 判断语句
# 其中elif是else if的缩写, 可以有多个elif, 但是只能有一个else
# else是可选的, 可以没有else
rain = True
if rain == True:
    print("下雨了, 出门记得带伞")
elif rain == False:
    print("没有下雨, 可以出去玩")
else:
    print("今天天气不错, 可以出去玩")
```

```python
# 循环语句 for 和 while
# while循环会先判断条件是否成立, 如果成立就会执行循环体, 然后再判断条件是否成立, 如果不成立就会跳出循环
# while最小执行circle是0次, 如果条件不成立就不会执行循环体
count = 0
while count < 5:
    print("count =", count)
    count += 1 # 等价于 count = count + 1
print("循环结束")

# for循环会先判断条件是否成立, 如果成立就会执行循环体, 然后再判断条件是否成立, 如果不成立就会跳出循环
# for循环最小执行circle是0次, 如果条件不成立就不会执行循环体
for i in range(5): # range(5)表示从0到4, 总共5个数, 也可以写成range(0, 5)
    print("i =", i)
```

```python
# 类数据结构的数据类型, 是否可重复, 是否可修改, 是否有序
# list 列表, 可重复, 可修改, 有序
# list中可以放任意类型的数据, 可以放list
list1 = []
list2 = [1,2,3,4,5,6,7,8,9]
list3 = ["Jacky", "Bob", "Nancy", "Nancy"]
print(list3[2])
print(len(list3))

# 元组(tuple), 可重复, 不可修改, 有序
tuple1 = ()
tuple2 = (1,2,3,4,5,6,7,8,9)
tuple3 = ("Jacky", "Bob", "Nancy", "Nancy")
print(tuple3[2])
print(len(tuple3))

# 集合(set), 不可重复, 可修改, 无序
set1 = set()
set2 = {1,2,3,4,5,6,7,8,9}
set3 = {"Jacky", "Bob", "Nancy", "Nancy"}
print(set3)
print(len(set3))

# 字典(dict), 不可重复的键, 可修改, 有序
dict1 = {}
dict2 = {"name": "Jacky", "age": 18}
print(len(dict2))
print(dict2["name"])
```

```python
# 数字炸弹
import random
secret_number = random.randint(1, 100) # 生成一个1到100之间的随机整数
guess_count = 0 # 猜测次数
while True:
    guess = int(input("请输入你猜的数字(1-100): "))
    guess_count += 1
    if guess < secret_number:
        print("你猜的数字太小了, 请再试一次")
    elif guess > secret_number:
        print("你猜的数字太大了, 请再试一次")
    else:
        print("恭喜你猜对了, 你一共猜了", guess_count, "次")
        break
```
